```python
import numpy as np

# 1. Definición de la función objetivo (Rosenbrock)
def f(x1, x2):
    return 100 * (x2 - x1*2)**2 + (1 - x1)**2

# 2. Gradiente
def grad_f(x1, x2):
    df_dx1 = 2 * (x1 - 1) - 400 * x1 * (x2 - x1**2)
    df_dx2 = 200 * (x2 - x1**2)
    return np.array([df_dx1, df_dx2])

# 3. Matriz Hessiana
def hess_f(x1, x2):
    d2f_dx12 = 2 - 400 * x2 + 1200 * x1**2
    d2f_dx1dx2 = -400 * x1
    d2f_dx22 = 200.0
    return np.array([[d2f_dx12, d2f_dx1dx2],
                     [d2f_dx1dx2, d2f_dx22]])

# --- Parámetros Iniciales ---
x = np.array([-1.2, 1.0])
alpha = 0.5  # <--- Agregamos una tasa de aprendizaje para dar pasos más controlados
iteraciones = 10  # <--- Aumentamos las iteraciones para ver la convergencia

print("--- Estado Inicial ---")
print(f"Iteración 0: x = {x}, f(x) = {f(x[0], x[1]):.4f}\n")

# --- Ejecución del método de Newton ---
for i in range(1, iteraciones + 1):
    g = grad_f(x[0], x[1])
    H = hess_f(x[0], x[1])

    # Resolver H * delta = -g
    delta = np.linalg.solve(H, -g)

    # Actualizar el punto usando la tasa de aprendizaje alpha
    x = x + alpha * delta

    print(f"--- Iteración {i} ---")
    print(f"Nuevo punto x: {x}")
    print(f"f(x) en el nuevo punto: {f(x[0], x[1]):.4f}\n")

    # Criterio de parada opcional: si el gradiente es casi cero, ya llegamos
    if np.linalg.norm(g) < 1e-5:
        print("¡El algoritmo ha convergido al mínimo!")
        break
```
